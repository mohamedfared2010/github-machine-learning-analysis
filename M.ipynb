```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
```

```python
# GitHub API URL
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
```

```python
# Read the data
data = pd.read_json(url)
```

```python
df = pd.DataFrame(data["items"].tolist())
```

```python
# Select needed columns
df = df[
    [
        "name",
        "owner",
        "language",
        "stargazers_count",
        "forks_count",
        "watchers_count",
        "open_issues_count",
        "created_at",
        "updated_at",
        "license"
    ]
]
```

```python
df["owner"] = df["owner"].str["login"]

df["license"] = df["license"].str["name"]

df["license"] = df["license"].fillna("No License")
```

```python
df.isnull().sum()
```

```python
# Handle missing values
df["language"] = df["language"].fillna("Unknown")
```

```python
df.duplicated().sum()
```

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
# Change dates
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])
```

```python
# Rename columns
df.rename(columns={
    "stargazers_count": "stars",
    "forks_count": "forks",
    "watchers_count": "watchers",
    "open_issues_count": "open_issues",
    "created_at": "created_date",
    "updated_at": "updated_date"
})
```

```python
# Save the file
df.to_csv("github_projects.csv")
```

```python
# Load the CSV
df = pd.read_csv("github_projects.csv")
```

```python
df.rename(columns={
    'stargazers_count': 'stars',
    'forks_count': 'forks',
    'watchers_count': 'watchers',
    'open_issues_count': 'open_issues',
    'created_at': 'created_date',
    'updated_at': 'updated_date'
})
```

```python
# Create database
conn = sqlite3.connect("github_projects.db")

df.to_sql("Repositories", conn, if_exists="replace")
```

```python
M = 'SELECT name, stars FROM Repositories WHERE stars > 10000'
pd.read_sql_query(M, conn)
```

```python
M = "SELECT name FROM Repositories WHERE name LIKE '%Machine%'"
pd.read_sql_query(M, conn)
```

```python
M = 'SELECT name, stars, forks FROM Repositories WHERE stars > 10000 AND forks > 1000'
pd.read_sql_query(M, conn)
```

```python
M = "SELECT name, language FROM Repositories WHERE language = 'Python' OR language = 'JavaScript'"
pd.read_sql_query(M, conn)
```

```python
M = "SELECT name, language FROM Repositories WHERE NOT language = 'Python'"
pd.read_sql_query(M, conn)
```

```python
M = 'SELECT name, stars FROM Repositories ORDER BY stars DESC LIMIT 10'
pd.read_sql_query(M, conn)
```

```python
M = 'SELECT COUNT(*) AS total_repositories FROM Repositories'
pd.read_sql_query(M, conn)
```

```python
M = 'SELECT AVG(stars) AS average_stars FROM Repositories'
pd.read_sql_query(M, conn)
```

```python
M = 'SELECT language, COUNT(*) AS total FROM Repositories GROUP BY language HAVING COUNT(*) > 5'
pd.read_sql_query(M, conn)
```
